```python
from phonemizer import phonemize
phonemes=phonemize(
    sample["text"],
    language="vi",
    backend="espeak",
    strip=True #remove space
)
print(phonemes)
```

```python
print(len(phonemes)) #num_embedding
```

```python
import torch
import torch.nn as nn
embedding_dims=[32, 64, 128, 256]
for dim in embedding_dims:
  embedding=nn.Embedding(
      num_embeddings=len(phonemes),
      embedding_dim=dim
  )
  print(f"dimension={dim}; embedding shape=", embedding.weight.shape)
```
